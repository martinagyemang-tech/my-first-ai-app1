## 1 · Lists — one box for many things 📚

A **list** holds several values in order, in square brackets. Your Week 3 chatbot's *entire memory*
will literally be one Python list — so let's get comfortable now.

```python
skills = ["Python basics", "the terminal", "GitHub"]

print(skills)
print(len(skills), "skills so far")
print("First:", skills[0])     # positions start at 0, not 1!
print("Last:", skills[-1])     # -1 means "count from the end"
```

**Lists grow.** `.append()` adds to the end — the same method that will add every chat message
to your bot's memory in Week 3:

```python
skills.append("AI API calls")     # coming in Week 2
skills.append("Streamlit apps")   # also Week 2
print(skills)
print(f"After Week 2 you'll have {len(skills)} skills 💪")
```

## 2 · Loops — do it for each one 🔁

A `for` loop runs the same code once **for each item** in a list. Watch the indentation:
the indented lines are "inside" the loop.

```python
for skill in skills:
    print(f"✅ I can use {skill}")

print("...and I'm just getting started!")   # not indented = runs once, after the loop
```

🧪 **TRY IT:** `range(n)` gives you the numbers 0 to n-1 — great for "do this n times":

```python
for day in range(1, 6):
    print(f"Week {day}: still showing up 🔥")
```

## 3 · Decisions — `if`, `elif`, `else` 🚦

Programs choose paths. The condition after `if` is anything that's `True` or `False`:

```python
hours_practiced = 4    # ✏️ change me and re-run!

if hours_practiced >= 5:
    print("🏆 Outstanding week!")
elif hours_practiced >= 2:
    print("💪 Solid — consistency beats intensity.")
else:
    print("🌱 Start small: even 30 minutes counts.")
```

**Comparison operators** you'll use daily: `==` equal (two! one `=` assigns), `!=` not equal,
`>` `<` `>=` `<=`, plus `and`, `or`, `not` to combine.

## 4 · Loops + decisions together = real programs

```python
test_scores = [45, 82, 67, 91, 38, 74]

passed = 0
for score in test_scores:
    if score >= 50:
        passed = passed + 1     # can also be written: passed += 1

print(f"{passed} out of {len(test_scores)} students passed")
print(f"Pass rate: {passed / len(test_scores) * 100:.0f}%")   # :.0f = round to whole number
```

## 5 · Dictionaries — labelled information 🏷️

A **dictionary** stores `key: value` pairs. This one matters enormously:
**every message you send to OpenAI next week is a dictionary** shaped exactly like this:

```python
# 👀 Remember this shape — you will type it 100 times in this course:
message = {"role": "user", "content": "What is the capital of Ghana?"}

print(message["role"])       # look up a value by its key
print(message["content"])
```

And a **conversation** is just a *list of dictionaries* — which you can now read fluently:

```python
conversation = [
    {"role": "system", "content": "You are a friendly assistant."},
    {"role": "user", "content": "What is the capital of Ghana?"},
    {"role": "assistant", "content": "The capital of Ghana is Accra."},
]

for msg in conversation:
    print(f"[{msg['role'].upper()}] {msg['content']}")
```

🤯 **You just read the exact data structure behind ChatGPT.** Week 3's chatbot memory is
literally this list, growing with `.append()` — two ideas you now own.

## 6 · Mini-project: study streak tracker

Everything combined. Run it, then make it yours:

```python
streak = ["Mon", "Tue", "Wed", "Fri", "Sat"]     # days you practiced this week
target = 5

print("📅 THIS WEEK'S PRACTICE LOG")
for day in ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]:
    if day in streak:                    # "in" checks membership in a list!
        print(f"  {day}: ✅")
    else:
        print(f"  {day}: ·")

if len(streak) >= target:
    print(f"\n🏆 {len(streak)}/{target} days — target smashed!")
else:
    print(f"\n💪 {len(streak)}/{target} days — {target - len(streak)} more to hit target.")
```
